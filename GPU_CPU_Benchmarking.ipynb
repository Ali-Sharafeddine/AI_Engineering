```python
import torch
import tqdm
import transformers
```

```python
from transformers import (
    pipeline,
    TextClassificationPipeline,
)

from tqdm import tqdm
import time
```

```python
def time_text_classifier(
    text_pipeline: TextClassificationPipeline,
    texts: list[str],
    batch_size: int = 1
) -> None :
  texts_generator = (t for t in texts) #comprehensive
  pipeline_iterable = tqdm(
      text_pipeline(
          texts_generator,
          batch_size = batch_size,
          truncation = True,
          max_length = 500
      ),
      total = len(texts)
  )
  for result in pipeline_iterable:
    pass
```

```python
model_name = "cardiffnlp/twitter-roberta-base-sentiment-latest"
sentiment_pipeline_cpu = pipeline(model = model_name, device = -1)
sentiment_pipeline_gpu = pipeline(model = model_name, device =  0)
```

```python
time_text_classifier(sentiment_pipeline_cpu, review_list[0:1000])
```

```python
time_text_classifier(sentiment_pipeline_gpu, review_list[0:1000])
```
